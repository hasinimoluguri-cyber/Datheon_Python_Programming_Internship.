```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
data = pd.read_excel("/content/sample transactions.xlsx")
```

```python
print("Dataset:")
print(data)
```

```python
# Split the single column into individual columns using comma as the delimiter
if "description,amount,category" in data.columns:
    data = data["description,amount,category"].str.split(",", expand=True)
    data.columns = ["description", "amount", "category"]

data = data.dropna(subset=["description", "category"])
```

```python
X = data["description"]
y = data["category"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)
```

```python
model = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("classifier", LogisticRegression(max_iter=1000))
])
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print("\nModel Accuracy:")
print(f"{accuracy_score(y_test, y_pred):.2%}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

```

```python
new_expenses = [
    "swiggy dinner",
    "uber ride",
    "electricity bill",
    "college books"
]

predictions = model.predict(new_expenses)

print("\nExpense Predictions:")
for description, category in zip(new_expenses, predictions):
    print(f"{description} -> {category}")
```
